**Soal 1 (OpenCV & Matplotlib):**
*   Baca gambar `orange.jpg` menggunakan OpenCV, ubah ruang warnanya ke RGB, lalu tampilkan menggunakan Matplotlib dengan ukuran plot $6 \times 6$.

```python

import cv2
import matplotlib.pyplot as plt

# Membaca gambar menggunakan OpenCV
img = cv2.imread('orange.jpg')

# Memeriksa apakah gambar berhasil dibaca
if img is None:
    raise FileNotFoundError("Gambar orange.jpg tidak ditemukan!")

# Mengubah warna gambar dari BGR menjadi RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Memperbesar gambar agar memenuhi ukuran cropping soal nomor 2
tinggi, lebar, kanal = img_rgb.shape

if tinggi < 600 or lebar < 800:
    skala = max(600 / tinggi, 800 / lebar)
    lebar_baru = int(lebar * skala)
    tinggi_baru = int(tinggi * skala)

    img_rgb = cv2.resize(
        img_rgb,
        (lebar_baru, tinggi_baru),
        interpolation=cv2.INTER_LINEAR
    )

# Menampilkan ukuran gambar
print("Ukuran gambar asli:", img.shape)
print("Ukuran gambar setelah resize:", img_rgb.shape)

# Menampilkan gambar menggunakan Matplotlib
plt.figure(figsize=(6, 6))
plt.imshow(img_rgb)
plt.title("Gambar Orange")
plt.axis("off")
plt.show()
```

**Penjelasan Soal 1:** `cv2.imread()` membaca gambar dalam format BGR. `cv2.cvtColor()` mengubahnya menjadi RGB agar warna sesuai ketika ditampilkan dengan Matplotlib. `figsize=(6, 6)` mengatur ukuran plot.

**Soal 2 (NumPy - Cropping):**
*   Lakukan pemotongan (cropping) gambar pada koordinat baris/y $200$ hingga $600$ dan kolom/x $400$ hingga $800$ menggunakan slicing array NumPy, lalu tampilkan hasilnya.



**Penjelasan Soal 2:** `img_rgb[200:600, 400:800]` mengambil baris 200 sampai sebelum 600 dan kolom 400 sampai sebelum 800. Hasilnya adalah potongan gambar dengan ukuran maksimum 400 × 400 piksel.

```python
import matplotlib.pyplot as plt

# Melakukan cropping menggunakan NumPy
crop = img_rgb[200:600, 400:800]

# Menampilkan hasil cropping
plt.figure(figsize=(6, 6))
plt.imshow(crop)
plt.title("Hasil Cropping")
plt.axis("off")
plt.show()

print("Ukuran hasil crop:", crop.shape)
```

**Soal 3 (Scikit-Image - Filtering):**
*   Konversi hasil potongan (crop) dari Soal 2 menjadi grayscale, lalu terapkan filter deteksi tepi Sobel (`filters.sobel`) dari Scikit-Image dan tampilkan hasilnya dengan peta warna (cmap) abu-abu (`gray`).

```python
import matplotlib.pyplot as plt
from skimage import color, filters

# Mengubah gambar hasil cropping menjadi grayscale
gray = color.rgb2gray(crop)

# Menerapkan filter Sobel
sobel_image = filters.sobel(gray)

# Menampilkan hasil filter Sobel
plt.figure(figsize=(6, 6))
plt.imshow(sobel_image, cmap='gray')
plt.title("Hasil Filter Sobel")
plt.axis("off")
plt.show()
```

**Penjelasan Soal 3:** Citra RGB hasil crop diubah menjadi grayscale agar menjadi satu kanal intensitas. `filters.sobel()` kemudian menghitung perubahan intensitas untuk menonjolkan bagian tepi objek. `cmap="gray"` menampilkan hasil dalam gradasi abu-abu.
