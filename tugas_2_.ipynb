````python

# Tugas 3 - Operasi Matematika

# Meminta pengguna memasukkan dua angka
angka1 = float(input("Masukkan angka pertama: "))
angka2 = float(input("Masukkan angka kedua: "))

# Melakukan operasi matematika
penjumlahan = angka1 + angka2
pengurangan = angka1 - angka2
perkalian = angka1 * angka2
pangkat = angka1 ** angka2

# Menampilkan hasil perhitungan
print("\n=== HASIL KALKULATOR ===")
print("Penjumlahan :", penjumlahan)
print("Pengurangan :", pengurangan)
print("Perkalian   :", perkalian)
print("Pangkat     :", pangkat)

# Memeriksa agar pembagian dengan nol tidak dilakukan
if angka2 != 0:
    pembagian = angka1 / angka2
    print("Pembagian   :", pembagian)
else:
    print("Pembagian   : Tidak bisa dibagi dengan nol")

# Meminta input panjang dan lebar persegi panjang
panjang = float(input("\nMasukkan panjang persegi panjang: "))
lebar = float(input("Masukkan lebar persegi panjang: "))

# Menghitung luas persegi panjang
luas = panjang * lebar

# Menampilkan hasil luas
print("Luas persegi panjang:", luas)
````

````python
# Tugas 4: Perulangan for dan while

# Daftar berisi lima nama mahasiswa.
nama_mahasiswa = ["Zulfa", "Azza", "Wira", "Khafidz", "Gabril"]

# Tampilkan daftar nama menggunakan for.
print("Daftar mahasiswa (for):")
for nama in nama_mahasiswa:
    print("-", nama)

# range(1, 11) menghasilkan angka 1 sampai 10.
print("\nAngka 1 sampai 10:")
for angka in range(1, 11):
    print(angka, end=" ")
print()

# Gunakan while dan len() untuk mengakses seluruh daftar nama.
print("\nDaftar mahasiswa (while):")
indeks = 0
while indeks < len(nama_mahasiswa):
    print("-", nama_mahasiswa[indeks])
    indeks += 1

# Tampilkan bilangan genap dari 2 sampai 20.
print("\nBilangan genap 2 sampai 20:")
bilangan = 2
while bilangan <= 20:
    print(bilangan, end=" ")
    bilangan += 2
print()
````

````python
# Tugas 5: Fungsi dan waktu
import datetime

# Fungsi ini menampilkan pesan dan waktu saat dipanggil.
def catat_waktu():
    waktu_sekarang = datetime.datetime.now()
    print("Aktivitas selesai")
    print("Waktu:", waktu_sekarang.strftime("%d-%m-%Y %H:%M:%S"))

# Catat waktu sebelum proses pengerjaan.
print("Mulai aktivitas belajar:")
catat_waktu()

# Simulasikan proses pengerjaan lima soal menggunakan for.
print("\nProses pengerjaan soal:")
for nomor_soal in range(1, 6):
    print("Mengerjakan soal nomor", nomor_soal)

# Catat waktu setelah proses selesai.
print("\nSelesai aktivitas belajar:")
catat_waktu()
````

````python
# Tugas proyek akhir: Sistem Rekap Nilai Mahasiswa
import datetime

# Tampilkan judul program.
print("=" * 36)
print("       REKAP NILAI MAHASISWA")
print("=" * 36)

# Minta identitas mahasiswa.
nama = input("Masukkan nama mahasiswa: ")
nim = input("Masukkan NIM mahasiswa: ")

# Minta nilai untuk tiga mata kuliah.
nama_mata_kuliah = ["Jarkom", "Kti", "Pengelolaan Citra"]
nilai = []
for mata_kuliah in nama_mata_kuliah:
    nilai_mata_kuliah = float(input(f"Masukkan nilai {mata_kuliah}: "))
    nilai.append(nilai_mata_kuliah)

# Hitung total dan rata-rata nilai.
total = sum(nilai)
rata_rata = total / len(nilai)

# Fungsi untuk menampilkan ringkasan nilai.
def tampilkan_hasil(total_nilai, rata_nilai):
    print("-" * 36)
    print("Total Nilai :", total_nilai)
    print("Rata-rata   :", rata_nilai)

# Fungsi untuk mencatat waktu penyelesaian program.
def catat_waktu():
    waktu_sekarang = datetime.datetime.now()
    print("Waktu       :", waktu_sekarang.strftime("%d-%m-%Y %H:%M:%S"))

# Tampilkan identitas dengan nama dalam huruf kapital.
print("\n" + "=" * 36)
print("       HASIL REKAP NILAI")
print("=" * 36)
print("Nama       :", nama.upper())
print("NIM        :", nim)
print("-" * 36)

# Gunakan for untuk menampilkan seluruh nilai mata kuliah.
for indeks in range(len(nama_mata_kuliah)):
    print(f"Nilai {nama_mata_kuliah[indeks]:<15}: {nilai[indeks]}")

# Gunakan while untuk menampilkan nomor urut mata kuliah.
print("\nNomor urut mata kuliah:")
nomor = 0
while nomor < len(nama_mata_kuliah):
    print(f"{nomor + 1}. {nama_mata_kuliah[nomor]}")
    nomor += 1

# Tampilkan total, rata-rata, status, dan waktu penyelesaian.
tampilkan_hasil(total, rata_rata)
print("-" * 36)
print("Status      : Rekap selesai")
catat_waktu()
print("=" * 36)
````
